## Using a MathFunction to verify that our algorithm is set up correctly

Trying to implement the formula which is defined [here](https://math.stackexchange.com/questions/154941/how-to-calculate-the-number-of-possible-connected-simple-graphs-with-n-labelle)

 

```kotlin
import main.dsl.expressions.ScalarAlgebra

val n = Variable("n").setTo(0)
val x = Variable("x").setTo(5)
val connectedGraphsWithXVertices = MathFunction(
    ScalarAlgebra.Sum(
        n,
        mathNum(0),
        mathNum(10000),
        {
            val numerator = 2.0.num() pow (n * (n - 1) / 2)
            (numerator / ScalarAlgebra.Factorial(n)) * (x pow n)
        }),
    "D(x)"
)

connectedGraphsWithXVertices.evaluate().also { println(it) }
connectedGraphsWithXVertices.display()
```

$$
D(x) = (n) \mapsto \sum_{n=0}^{10000}{\frac{2^{\frac{n \cdot (n - 1)}{2}}}{n!} \cdot x^{n}} = 10001.0
$$
